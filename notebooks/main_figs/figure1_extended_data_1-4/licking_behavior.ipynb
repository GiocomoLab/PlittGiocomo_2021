## Average lick rates per mouse

```python
def build_lick_matrix(sess_dict):
    
    def _single_mouse(mouse, sess_list):
        
        lick_mu, morphs, rzone_pos = [], [], []
        for sess_deets in sess_list:
            sess = m.sess.CA1MorphSession.from_nwb(mouse, 'testing', sess_deets['day'])
            lick_mu.append(sess.trial_matrices['licks'])
            morphs.append(sess.trial_info['morphs'][:,np.newaxis])
            rzone_pos.append(sess.trial_info['rzone_entry'][:,np.newaxis])

        return np.vstack(lick_mu), np.vstack(morphs), np.vstack(rzone_pos)
            
            
    L = {}
    for mouse, metadata in sess_dict.items():
        sess_list = metadata['test_sessions']
        licks, morphs, rzone_pos = _single_mouse(mouse, sess_list)
        L[mouse] = {
            'licks': licks,
            'morphs': morphs,
            'rzone_pos': rzone_pos
        }
        
    return L
```

```python
rare_lick = build_lick_matrix(m.mouse_metadata.rare_sessions)
freq_lick = build_lick_matrix(m.mouse_metadata.frequent_sessions)

all_lick = {**rare_lick, **freq_lick}
```

```python
f,ax = plt.subplots(1,2,figsize=[10,5])
nbins = 45
cmaps = ['copper','cividis']



for i,L in enumerate([rare_lick,freq_lick]):
    nmice = len(L.keys())
    cm  = plt.cm.get_cmap(cmaps[i])
    
    for z, mouse in enumerate(L.keys()):
        licks = L[mouse]['licks']
        
        morphs = L[mouse]['morphs']
        print(licks.shape)
        mu = licks.mean(axis=0)
        sem = sp.stats.sem(licks,axis=0)
        
        
        ax[0].fill_between(np.linspace(0,450,num=nbins),mu-sem,y2=mu+sem,alpha=.3,color=cm(float(z/nmice)))
        
        
        licks_n = np.copy(licks)
        licks_n/=licks_n.ravel().mean()
        mu = licks_n.mean(axis=0)
        sem = sp.stats.sem(licks_n,axis=0)
        
        ax[1].fill_between(np.linspace(0,450,num=nbins),mu-sem,y2=mu+sem,alpha=.3,color=cm(float(z/nmice)))
ax[0].fill_betweenx([0,.6],225,400,color='black',alpha=.3,zorder=0)
ax[1].fill_betweenx([0,8],225,400,color='black',alpha=.3,zorder=0)
ax[0].set_ylim([-.01,.6])
ax[1].set_ylim([-.01,8])


ax[0].set_ylabel('Licks/Sec')
ax[1].set_ylabel('Lick Rate (norm.)')

for a in ax.flatten():
    a.set_xlabel('Position (cm)')
    a.spines['top'].set_visible(False)
    a.spines['right'].set_visible(False)
```

## Average lick rates by position of reward cue

```python

def pos_binned_licks(lick_mat,rpos,rz_edges=[220,260,290,320,380,481]):
    rpos[np.isnan(rpos)]=480
    rz_bins = np.digitize(rpos,rz_edges)-1
    mu_mat = np.zeros([len(rz_edges)-1,lick_mat.shape[1]])
    for b in range(len(rz_edges)-1):
        mu_mat[b,:] = lick_mat[rz_bins==b,:].mean(axis=0)
    return mu_mat

rz_edges=[220,260,290,320,380,481]
nmice = len(all_lick.keys())
# f,ax = plt.subplots()

MU_MAT=[]
for z, mouse in enumerate(all_lick.keys()):
    licks = all_lick[mouse]['licks']
    licks_n = np.copy(licks)
    licks_n/=licks.ravel().mean()
    
    rpos = all_lick[mouse]['rzone_pos'].ravel()
    _mu_mat = pos_binned_licks(licks_n,rpos,rz_edges=rz_edges)
    MU_MAT.append(_mu_mat)
MU_MAT = np.squeeze(np.array(MU_MAT))
print(MU_MAT.shape)



f,ax = plt.subplots()
ax.fill_betweenx([0,10],225,400,color='black',alpha=.3,zorder=0)

lr,sem = np.nanmean(MU_MAT,axis=0),sp.stats.sem(MU_MAT,axis=0,nan_policy='omit')
for b in range(len(rz_edges)-1):
    if b==4:
        ax.fill_between(np.linspace(0,450,num=nbins),lr[b,:]-sem[b,:],lr[b,:]+sem[b,:],color='black',alpha=.3)
    else:
        ax.fill_between(np.linspace(0,450,num=nbins),lr[b,:]-sem[b,:],lr[b,:]+sem[b,:],color=plt.cm.viridis(rz_edges[b+1]/rz_edges[4]),alpha=.3)
        
ax.set_ylabel('Norm. lick rate')
ax.set_xlabel('position (cm)')
ax.set_title('Mean lick rate binned by reward position')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
```
