```python
# imports
import morph_analyses as m
from matplotlib import pyplot as plt
import numpy as np
import pathlib

%matplotlib inline

%load_ext autoreload
%autoreload 2
```

```python
mouse = 'F2'
sess_deets = m.mouse_metadata.frequent_sessions[mouse]['test_sessions'][-1]
sess = m.sess.CA1MorphSession.from_nwb(mouse, 'testing', sess_deets['day'])
```

```python
ops = sess.s2p_ops
stat_cells = sess.s2p_stats
iscell = sess.iscell
```

```python
print(ops.keys())
f,ax = plt.subplots(figsize=[10,10])
ax.imshow(ops['meanImg'],aspect=796./512,cmap='Greys_r')

# plot ROIs
roi_im = np.zeros((ops['Ly'], ops['Lx']))
roi_im[:]=np.nan

for n in range(0,len(stat_cells)):
    ypix = stat_cells[n]['ypix']
    xpix = stat_cells[n]['xpix']
    roi_im[ypix,xpix] = n
    
ax.imshow(roi_im,cmap='hsv',alpha=.3,aspect=796/512.)
```
