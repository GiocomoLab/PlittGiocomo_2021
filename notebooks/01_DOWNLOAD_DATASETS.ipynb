### Download frame grabs from unity for morph axis definition



```python

import requests
import subprocess
import os

article_id = "31968255"

# Construct the API URL for the article details (using API v2)
api_url = f"https://api.figshare.com/v2/articles/{article_id}"
article = requests.get(api_url).json()

# Create directory if it doesn't exist
output_dir = "../data"
os.makedirs(output_dir, exist_ok=True)

for file in article["files"]:
    # Using file["name"] ensures we don't overwrite files
    dest_path = os.path.join(output_dir, file["name"])
    print(f'Downloading {file["name"]}...')
    
    # Added -L to follow redirects, which Figshare requires
    subprocess.run(["curl", "-L", file["download_url"], "-o", dest_path], check=True)
```
